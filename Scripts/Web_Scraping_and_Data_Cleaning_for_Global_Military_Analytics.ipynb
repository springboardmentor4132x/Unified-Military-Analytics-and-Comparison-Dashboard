```python
url=[]
names=[]
for link,name in Dict.items():
    url.append(link)
    names.append(name)
```

```python
import requests
from bs4 import BeautifulSoup
import re
```

```python
final_result=[]
for link,name in Dict.items():
    u=link
    response=requests.get(u)
    soup = BeautifulSoup(response.text, 'html.parser')
    all_links=soup.find_all('a')
    specific_urls=[]
    for link in all_links:
        href = link.get("href")
        if "country_id" in href:
            specific_urls.append(link)

    for record in specific_urls:
        country_div = record.find("div", class_="longFormName")
        value_div = record.find("div", class_="valueContainer")

        if not country_div or not value_div:
            continue

        country = country_div.text.strip()
        value = value_div.text.strip()
        regex = "[^0-9]"
        value=(re.sub(regex, "", value))

        final_result.append({
            "country": country,
            "metric": name,
            "value": int(value)
        })

```

```python
print(final_result)
```

```python
import pandas as pd

df = pd.DataFrame(final_result)
df.head()
```

```python
df = df.pivot(
    index="country",
    columns="metric",
    values="value"
)
```
