```python
import pandas as pd
df = pd.read_csv(r"/content/Output_for_Milestone1.csv")
print("Columns count:", len(df.columns))

columns_to_drop = [
    "oil_consumption_bbl",
    "oil_production_bbl",
    "natural_gas_consumption_cum",
    "natural_gas_production_cum",
    "coal_consumption_mt",
    "coal_production_cum",
    "proven_oil_reserves_bbl",
    "proven_natural_gas_reserves_cum",
    "proven_coal_reserves_cum",
    "labour_force",
    "external_debt_usd",
    "foreign_exchange_and_gold_reserves_usd",
    "total_merchant_marine_fleet",
    "mine_warfare_craft"
]

df.drop(columns=[col for col in columns_to_drop if col in df.columns],inplace=True)

print("Unnecessary columns removed successfully.")
print("Final columns count:", len(df.columns))
```

```python
df.head()
```

```python
#Power Index Rank Gap
df['Power_Score']=df['active_personnel']+df['tanks']+df['total_military_aircraft']+df['total_naval_fleet']
df["Power_Rank"] = df["Power_Score"].rank(ascending=False, method="min")
df["Power_Index_Rank_Gap"] = df["Power_Rank"] - 1
```

```python
df.columns
```

```python
#Assets per Capita
df["Total_Assets"] = (df["tanks"]+df["total_military_aircraft"]+df["total_naval_fleet"])
df["Assets_Per_Capita"] = df["Total_Assets"]/df["total_population"]
```

```python
#Budget-to-GDP Ratio
df["Budget_to_GDP_Ratio"] = (df["defense_budget_usd"] / df["purchasing_power_parity_usd"])
```
