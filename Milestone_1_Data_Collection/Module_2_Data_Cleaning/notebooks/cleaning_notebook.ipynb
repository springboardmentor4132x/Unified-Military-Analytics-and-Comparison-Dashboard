# Module 2: Data Cleaning - Military Metrics

This notebook demonstrates how to clean raw military data using the cleaner script.

## Overview
- Load raw scraped data
- Standardize column names (snake_case)
- Remove special characters
- Handle missing values
- Convert to numeric types
- Remove duplicates
- Generate cleaning report

## 1. Load Dependencies

```python
import sys
sys.path.append('../')

from clean_data import MilitaryDataCleaner
import pandas as pd
import numpy as np
from datetime import datetime

print("✅ Dependencies loaded successfully")
```

## 2. Initialize Cleaner

```python
# Initialize cleaner with configuration
cleaner = MilitaryDataCleaner(
    config_file='data_mapping.json',
    log_file='cleaning_log.txt'
)

print("✅ Data Cleaner initialized")
print(f"Configuration loaded from: data_mapping.json")
```

## 3. Load Raw Data

```python
# Load raw data
raw_df = pd.read_csv('data/raw/military_raw_data.csv')

print(f"📊 Raw Data Shape: {raw_df.shape}")
print(f"\n📋 Raw Data Sample:")
print(raw_df.head())
print(f"\n⚠️ Missing Values in Raw Data:")
print(raw_df.isnull().sum())
```

## 4. Execute Cleaning Pipeline

```python
# Run cleaning pipeline
print("🧹 Starting data cleaning pipeline...\n")
cleaned_df = cleaner.clean(raw_df)
print("✅ Cleaning completed!")
```

## 5. Validate Cleaned Data

```python
print(f"📊 Cleaned Data Shape: {cleaned_df.shape}")
print(f"\n📋 Cleaned Data Sample:")
print(cleaned_df.head())
print(f"\n✅ Column Names (standardized):")
print(list(cleaned_df.columns))
print(f"\n📈 Data Types:")
print(cleaned_df.dtypes)
```

## 6. Quality Metrics

```python
# Calculate quality metrics
total_cells = cleaned_df.shape[0] * cleaned_df.shape[1]
missing_cells = cleaned_df.isnull().sum().sum()
missing_rate = (missing_cells / total_cells) * 100
duplicates = cleaned_df.duplicated(subset=['country']).sum()

print("\n" + "="*50)
print("DATA QUALITY REPORT")
print("="*50)
print(f"✅ Rows: {cleaned_df.shape[0]}")
print(f"✅ Columns: {cleaned_df.shape[1]}")
print(f"⚠️ Missing Values: {missing_cells} ({missing_rate:.2f}%)")
print(f"🔄 Duplicates Removed: {duplicates}")
print(f"📊 Numeric Columns: {len(cleaned_df.select_dtypes(include=[np.number]).columns)}")
print(f"📝 Text Columns: {len(cleaned_df.select_dtypes(include=['object']).columns)}")
print("="*50)

if missing_rate < 2:
    print("✅ SUCCESS - Data quality acceptable (< 2% missing)")
else:
    print(f"⚠️ WARNING - Missing data rate: {missing_rate:.2f}%")
```

## 7. Save Cleaned Data

```python
# Save cleaned data
output_file = 'data/processed/military_cleaned.csv'
cleaned_df.to_csv(output_file, index=False)

print(f"💾 Cleaned data saved to: {output_file}")
print(f"✅ Ready for next phase: KPI Engineering")
```
