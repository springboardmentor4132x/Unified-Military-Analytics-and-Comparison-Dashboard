# Module 1: Web Scraping - Military Metrics

This notebook demonstrates how to scrape military data from GlobalFirepower.com using the scraper script.

## Overview
- Load configuration from `scraper_config.json`
- Scrape military metrics for 140+ countries
- Implement retry logic and error handling
- Validate data quality
- Generate scraping report

## 1. Setup and Configuration

```python
import sys
sys.path.append('../')

from scrape_military_metrics import MilitaryMetricsScraper
import json
import pandas as pd
from datetime import datetime

# Initialize scraper
scraper = MilitaryMetricsScraper(
    config_file='scraper_config.json',
    output_file='data/raw/military_raw_data.csv',
    log_file='scraping_log.txt'
)

print("✅ Scraper initialized successfully")
print(f"Base URL: {scraper.base_url}")
print(f"Timeout: {scraper.timeout}s")
print(f"Retry attempts: {scraper.retry_attempts}")
```

## 2. Load Country URLs

```python
# Load URLs from file
urls = scraper.load_urls_from_file('links_for_military_data.txt')

print(f"Loaded {len(urls)} URLs")
print(f"\nFirst 5 URLs:")
for url in urls[:5]:
    print(f"  • {url}")
```

## 3. Execute Scraping

```python
# Run the scraper
print("🔍 Starting scraping process...")
scraper.scrape_all_countries(urls)
print("✅ Scraping completed!")
```

## 4. Validate Output

```python
# Load and validate output
df = pd.read_csv('data/raw/military_raw_data.csv')

print(f"📊 Data Summary:")
print(f"  Shape: {df.shape} (rows, columns)")
print(f"  Countries: {len(df)}")
print(f"  Missing values: {df.isnull().sum().sum()} ({(df.isnull().sum().sum()/(df.shape[0]*df.shape[1])*100):.2f}%)")
print(f"\n📋 Sample data:")
print(df.head())
print(f"\n📈 Data types:")
print(df.dtypes)
```

## 5. Success Report

```python
# Calculate success metrics
success_rate = (len(df) / len(urls)) * 100 if len(urls) > 0 else 0
missing_rate = (df.isnull().sum().sum() / (df.shape[0] * df.shape[1]) * 100)

print("\n" + "="*50)
print("SCRAPING SUCCESS REPORT")
print("="*50)
print(f"✅ Success Rate: {success_rate:.2f}%")
print(f"📊 Countries Scraped: {len(df)}/{len(urls)}")
print(f"📈 Indicators per Country: {len(df.columns)-1}")
print(f"⚠️ Missing Data: {missing_rate:.2f}%")
print(f"💾 Output File: data/raw/military_raw_data.csv")
print(f"📝 Log File: scraping_log.txt")
print("="*50)

if success_rate >= 95:
    print("✅ SUCCESS - Ready for next phase (Data Cleaning)")
else:
    print(f"⚠️ WARNING - Success rate below 95%. Review logs.")
```
