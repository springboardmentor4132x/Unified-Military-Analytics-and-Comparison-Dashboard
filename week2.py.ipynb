```python
import pandas as pd
import numpy as np
import re
```

```python
INPUT_FILE = "military pivot data.csv"
OUTPUT_FILE = "military cleaned.csv"
```

```python
print("Loading military pivot data...")
data_frame = pd.read_csv(INPUT_FILE)
print("Rows loaded:", len(data_frame))
print("Columns found:", len(data_frame.columns))
```

```python
def is_country_column(column_name):
    """
    Checks whether the column represents country names.
    """
    return column_name.lower() in ["country", "country_name", "nation"]

def normalize_numeric_text(value):
    """
    Removes commas, percentage symbols, and extra characters,
    leaving only numbers, minus sign, and decimal points.
    """
    if pd.isna(value):
        return np.nan

    value = str(value).strip()
    value = re.sub(r"[,%+]", "", value)
    value = re.sub(r"[^\d.-]", "", value)

    return value
```

```python
print("Cleaning numeric values...")

metric_columns = []
for col in data_frame.columns:
    if not is_country_column(col):
        metric_columns.append(col)

for col in metric_columns:
    data_frame[col] = data_frame[col].apply(normalize_numeric_text)
```

```python
print("Converting values to numeric format...")

for col in metric_columns:
    data_frame[col] = pd.to_numeric(data_frame[col], errors="coerce")
```

```python
numeric_columns = data_frame.select_dtypes(include=["int64", "float64"]).columns

missing_summary = data_frame[numeric_columns].isna().sum()
print("Missing values before filling:")
print(missing_summary)
```

```python
for col in numeric_columns:
    median_value = data_frame[col].median()
    data_frame[col] = data_frame[col].fillna(median_value)
```

```python
final_missing = data_frame[numeric_columns].isna().sum().sum()
print("Remaining missing numeric values after fill:", final_missing)
```

```python
data_frame.to_csv(OUTPUT_FILE, index=False)

print("Data cleaning process completed successfully.")
print("Output file created:", OUTPUT_FILE)
```
