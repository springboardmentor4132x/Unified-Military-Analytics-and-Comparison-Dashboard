# Data Cleaning and Structuring
## Milestone 1 - Module 2

1. Load raw CSV
2. Remove commas, %, $, special symbols
3. Convert to numeric formats
4. Handle missing/null values
5. Save cleaned dataset

```python
import pandas as pd
import os


RAW_DATA = "../data/raw/military_raw_data.csv"
CLEAN_DATA = "../data/military_cleaned.csv"


os.makedirs(os.path.dirname(CLEAN_DATA), exist_ok=True)


df = pd.read_csv(RAW_DATA, encoding="utf-8")
print(f"Loaded {len(df)} countries with {len(df.columns)} columns")
```

```python
# head values
df.head(3)

```

```python

metric_columns = [col for col in df.columns if col != "country" and not col.endswith("_rank")]
rank_columns = [col for col in df.columns if col.endswith("_rank")]


print(f"Metric columns (to clean): {len(metric_columns)}")
```

```python
#Cleaning metric columns


for col in metric_columns:
    if col in df.columns:
        
        df[col] = df[col].astype(str)
        
       
        df[col] = df[col].str.replace("$", "", regex=False)
        
       
        df[col] = df[col].str.replace(",", "", regex=False)
        
       
        df[col] = df[col].str.replace("+", "", regex=False)
        
       
        df[col] = df[col].str.replace("%", "", regex=False)
        
        
        df[col] = df[col].str.replace("km", "", regex=False, case=False)
        df[col] = df[col].str.replace("bbl", "", regex=False, case=False)
        df[col] = df[col].str.replace("cum", "", regex=False, case=False)
        df[col] = df[col].str.replace("mt", "", regex=False, case=False)
        df[col] = df[col].str.replace("usd", "", regex=False, case=False)
        
        
        df[col] = df[col].str.replace("\n", " ", regex=False)
        df[col] = df[col].str.replace("\r", " ", regex=False)
        df[col] = df[col].str.strip()
        
        
        df[col] = pd.to_numeric(df[col], errors="coerce")
        
        


for col in rank_columns:
    if col in df.columns:
        
        df[col] = pd.to_numeric(df[col], errors="coerce").astype("Int64")
        
```

```python
missing_summary = df[metric_columns].isna().sum()

missing_summary = missing_summary[missing_summary > 0].sort_values(ascending=False)

missing_summary.head(20)
# df.shape   (145, 109)
 
```

```python
# Drop columns with missing data

print("DROPPING COLUMNS WITH MISSING DATA")


# List of columns to drop (including their rank columns)
columns_to_drop = [
    'natural_gas_production_by_country',
    'natural_gas_production_by_country_rank',
    'natural_gas_consumption_by_country',
    'natural_gas_consumption_by_country_rank',
    'coal_production_by_country',
    'coal_production_by_country_rank',
    'proven_natural_gas_reserves_by_country',
    'proven_natural_gas_reserves_by_country_rank',
    'proven_coal_reserves_by_country',
    'proven_coal_reserves_by_country_rank',
    'navy_force_by_tonnage',
    'navy_force_by_tonnage_rank'
]



# Drop these columns
df = df.drop(columns=columns_to_drop, errors='ignore') # dropped 12 columns

print(f"Remaining columns: {len(df.columns)}")
df.shape
```

```python
# saving cleaned data

df.to_csv(CLEAN_DATA, index=False, encoding="utf-8")
print(f" Saved {len(df)} countries with {len(df.columns)} columns")
```
