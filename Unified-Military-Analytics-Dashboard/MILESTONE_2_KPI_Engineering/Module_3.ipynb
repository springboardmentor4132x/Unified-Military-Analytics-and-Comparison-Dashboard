```python
import pandas as pd
import numpy as np

df = pd.read_excel("global_firepower_complete_dataset.xlsx")

df["total_military_assets"] = (
    df["total_military_aircraft"] +
    df["tanks"] +
    df["total_naval_fleet"]
)

df["derived_power_score"] = (
    df["total_military_assets"] * 0.5 +
    df["active_personnel"] * 0.3 +
    df["defense_budget_usd"] * 0.2
)

df["power_index_rank"] = df["derived_power_score"].rank(
    ascending=False,
    method="dense"
)

df["power_index_rank_gap"] = (
    df["power_index_rank"] -
    df["power_index_rank"].mean()
)

df["assets_per_capita"] = (
    df["total_military_assets"] /
    df["total_population"]
)

df["defense_budget_to_gdp_ratio"] = (
    df["defense_budget_usd"] /
    df["purchasing_power_parity_usd"]
)

print("✅ Required KPIs created successfully")
```

```python

df["spending_per_soldier"] = (
    df["defense_budget_usd"] /
    df["active_personnel"]
)

df["assets_per_100k_population"] = (
    df["assets_per_capita"] * 100000
)

df["budget_per_asset"] = (
    df["defense_budget_usd"] /
    df["total_military_assets"]
)

df["military_strength_index"] = (
    (df["total_military_assets"] * 0.4) +
    (df["active_personnel"] * 0.3) +
    (df["defense_budget_usd"] * 0.3)
)
```

```python
import pandas as pd



NATO = {
    "United States", "Canada", "United Kingdom", "France", "Germany",
    "Italy", "Spain", "Netherlands", "Belgium", "Norway", "Denmark",
    "Poland", "Turkey", "Greece", "Portugal", "Czech Republic",
    "Hungary", "Romania", "Bulgaria", "Slovakia", "Slovenia",
    "Estonia", "Latvia", "Lithuania", "Finland", "Sweden", "Croatia",
    "Albania", "North Macedonia", "Luxembourg", "Iceland", "Montenegro"
}

BRICS = {
    "Brazil", "Russia", "India", "China", "South Africa",
    "Saudi Arabia", "United Arab Emirates", "Iran",
    "Egypt", "Ethiopia"
}


df["is_nato"] = df["country"].apply(lambda x: 1 if x in NATO else 0)
df["is_brics"] = df["country"].apply(lambda x: 1 if x in BRICS else 0)


def classify_alliance(country):
    if country in NATO and country in BRICS:
        return "NATO & BRICS"
    elif country in NATO:
        return "NATO"
    elif country in BRICS:
        return "BRICS"
    else:
        return "Non-Aligned"

df["alliance_group"] = df["country"].apply(classify_alliance)



print(df["alliance_group"].value_counts())
```

```python

df.to_excel(
    "military_all_data_with_kpis_and_alliances.xlsx",
    index=False
)

kpi_alliance_cols = [
    "country",
    "iso",
    "alliance_group",
    "is_nato",
    "is_brics",
    "power_index_rank",
    "power_index_rank_gap",
    "assets_per_capita",
    "defense_budget_to_gdp_ratio",
    "military_strength_index"
]

df[kpi_alliance_cols].to_excel(
    "military_kpi_metrics_with_alliances.xlsx",
    index=False
)

print("✅ All 140+ countries classified and saved successfully")
```
