```python
import pandas as pd
import re

# Load data
df = pd.read_csv("military_raw_data.csv")
```

```python
print(df.dtypes)
```

```python
missing_metrics = df.isna().sum()
print("\nMissing metric counts per column:")
print(missing_metrics)
```

```python
# Clean text: remove commas, %, +, special symbols
def clean_value(val):
    if pd.isna(val):
        return None
    val = str(val)
    val = re.sub(r"[,+%]", "", val)
    val = re.sub(r"[^0-9.\-eE]", "", val)
    try:
        return float(val)
    except:
        return None

for col in df.columns:
    if col != "Country":
        df[col] = df[col].apply(clean_value)
```

```python
# Standardize column names
def to_snake_case(col_name):
    col_name = col_name.strip().replace(" ", "_").replace("-", "_").lower()
    col_name = re.sub(r"[^a-z0-9_]", "", col_name)
    return col_name
```

```python
# Check missing values per column
missing_counts = df.isna().sum()

# Filter columns that have at least one missing value
columns_with_missing = missing_counts[missing_counts > 0]

print("Columns with missing values and their counts:")
print(columns_with_missing)
```

```python
# Replace all missing values with 0
df.fillna(0, inplace=True)

# Verify missing values after replacement
print("Missing values after replacement:")
print(df.isna().sum())


```

```python
int_columns = [
    "total_population",
    "total_military_manpower",
    "fit_for_service",
    "population_reaching_military_age_annually",
    "active_personnel",
    "reserve_personnel",
    "paramilitary",
    "total_military_aircraft",
    "fighter_aircraft",
    "attack_aircraft",
    "transport_aircraft",
    "trainer_aircraft",
    "special_mission_aircraft",
    "tanker_aircraft",
    "total_military_helicopters",
    "attack_helicopters",
    "tanks",
    "armored_fighting_vehicles",
    "self_propelled_artillery",
    "towed_artillery",
    "rocket_projectors",
    "total_naval_fleet",
    "aircraft_carriers",
    "helicopter_carriers",
    "submarines",
    "destroyers",
    "frigates",
    "corvettes",
    "coastal_patrol_craft",
    "mine_warfare_craft",
    "total_serviceable_airports",
    "major_ports_and_terminals",
    "total_merchant_marine_fleet"
]
df[int_columns] = df[int_columns].astype(int)
```

```python
# Save updated cleaned data
df.to_csv("military_cleaned.csv", index=False)
```
